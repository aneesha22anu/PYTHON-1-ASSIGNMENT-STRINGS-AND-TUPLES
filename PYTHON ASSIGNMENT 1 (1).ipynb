###String Concatenation

   

```python
str1 = "Hello "
str2 = input("Enter your Name: ")

result = str1 + str2
print(result)
```

```python
str1 = "Hello "
str2 = input("Enter your Name: ")

result = str1 + str2
print(result)

str3 = ", welcome to Python programming"

result = result + str3
print(result)
```

## String Slicing and Indexing

```python
str1 = "Hello "
str2 = input("Enter your Name: ")

result = str1 + str2
result = result + ", welcome to Python programming"

print("First character:", result[0])
print("Last character:", result[-1])
print("First 5 characters:", result[:5])
print("Last 11 characters:", result[-11:])
print("Reverse string:", result[::-1])
print("Python:", result[18:24])
```
